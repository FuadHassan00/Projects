```python
import requests
import os
from dotenv import load_dotenv
from pathlib import Path
os.chdir(Path.home() / "Downloads/github-repos/Projects/Flight-Agent")
```

```python
# Load .env
load_dotenv(".env", override=True)

# Retrieve API key
api_key = os.getenv("SEATS_AERO_API_KEY")
```

```python
api_key
```

```python
url = "https://seats.aero/partnerapi/search"


headers = {
    "accept": "application/json",
    "Partner-Authorization": api_key.strip()
}

params = {
    "origin_airport": "CLT,JFK,EWR",
    "destination_airport": "PRG",
    "start_date": "2026-12-26",
    "end_date": "2026-12-30",
    "take": 100,
    "order_by": "lowest_mileage",
    "include_trips": True
}

response = requests.get(
    url,
    headers=headers,
    params=params,
    timeout=30
)

print(response.status_code)
print(response.json())
```

```python
import json
data = response.json()

print(json.dumps(data, indent=4))
```

```python
import pandas as pd

output = []

for flight in data.get("data", []):
    output.append({
        "date": flight["Date"],
        "program": flight["Source"],
        "economy_points": int(flight["YMileageCost"]),
        "economy_taxes": flight["YTotalTaxes"] / 100,
        "premium_economy_points": int(flight["WMileageCost"]),
        "premium_economy_taxes": flight["WTotalTaxes"] / 100,
        "business_points": int(flight["JMileageCost"]),
        "business_taxes": flight["JTotalTaxes"] / 100,
        "first_class_points": int(flight["FMileageCost"]),
        "first_class_taxes": flight["FTotalTaxes"] / 100,
        "currency": flight["TaxesCurrency"]
    })

df = pd.DataFrame(output)
```
